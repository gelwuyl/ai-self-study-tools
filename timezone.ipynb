```python
from datetime import datetime
from zoneinfo import ZoneInfo

# Get current date and time in Singapore (Asia/Singapore timezone)
current_sg_datetime = datetime.now(ZoneInfo("Asia/Singapore"))
print(f"Current date time in Singapore: {current_sg_datetime}")
```

```python
# get the current date time in Malaysia
current_my_datetime = datetime.now(ZoneInfo("Asia/Kuala_Lumpur"))
print(f"Current date time in Malaysia: {current_my_datetime}")
```

```python
# get the current date time in Japan
current_jp_datetime = datetime.now(ZoneInfo("Asia/Tokyo"))
print(f"Current date time in Japan: {current_jp_datetime}")
```

```python
# get the current date time in Cambodia
current_kh_datetime = datetime.now(ZoneInfo("Asia/Phnom_Penh"))
print(f"Current date time in Cambodia: {current_kh_datetime}")
```

```python
# get the current date time in Hong Kong, India
current_hk_datetime = datetime.now(ZoneInfo("Asia/Hong_Kong"))
print(f"Current date time in Hong Kong: {current_hk_datetime}")
current_in_datetime = datetime.now(ZoneInfo("Asia/Kolkata"))
print(f"Current date time in India: {current_in_datetime}")
```

```python
# get current date tiem in Australia
current_au_datetime = datetime.now(ZoneInfo("Australia/Sydney"))
print(f"Current date time in Australia: {current_au_datetime}")
```

```python
# get current date time in New Zealand
current_nz_datetime = datetime.now(ZoneInfo("Pacific/Auckland"))
print(f"Current date time in New Zealand: {current_nz_datetime}")
```

```python
# combine all the above into one string of code
timezones = {
    "Japan": "Asia/Tokyo",
    "Cambodia": "Asia/Phnom_Penh",
    "Hong Kong": "Asia/Hong_Kong",
    "India": "Asia/Kolkata",
    "Australia": "Australia/Sydney",
    "New Zealand": "Pacific/Auckland"
}
# display the output of the respective date and time
for country, tz in timezones.items():
    dt = datetime.now(ZoneInfo(tz))
    print(f"Current date time in {country}: {dt}")
```
